## 2. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print('Libraries imported successfully!')
```

## 3. Load Dataset

```python
df = pd.read_csv('dataset/sales_data.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

## 4. Data Understanding

```python
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nSummary statistics:')
print(df.describe())
```

## 5. Data Quality Checks

```python
print('Missing values:')
print(df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
print('\nUnique products:', df['Product'].unique())
print('Unique regions:', df['Region'].unique())
```

## 6. Data Cleaning

```python
df = df.drop_duplicates().copy()
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df = df.dropna(subset=['Sales','Quantity'])
print('Cleaned shape:', df.shape)
```

## 7. Data Manipulation

```python
print('Total Sales:', df['Sales'].sum())
print('Average Sale:', round(df['Sales'].mean(),2))
print('Total Quantity:', df['Quantity'].sum())
```

```python
product_analysis = df.groupby('Product').agg(
    Total_Sales=('Sales','sum'),
    Total_Quantity=('Quantity','sum'),
    Average_Sales=('Sales','mean')
).sort_values('Total_Sales', ascending=False)
product_analysis
```

```python
region_analysis = df.groupby('Region').agg(
    Total_Sales=('Sales','sum'),
    Total_Quantity=('Quantity','sum')
).sort_values('Total_Sales', ascending=False)
region_analysis
```

## 8. Exploratory Data Analysis

```python
monthly_sales = df.groupby(df['Date'].dt.to_period('M'))['Sales'].sum()
print(monthly_sales)
```

```python
print(df[['Quantity','Sales']].corr())
```

## 9. Data Visualization

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
plt.bar(region_sales.index, region_sales.values)
plt.title('Total Sales by Region')
plt.xlabel('Region'); plt.ylabel('Sales')
plt.show()
```

```python
product_sales = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
plt.bar(product_sales.index, product_sales.values)
plt.title('Total Sales by Product')
plt.xlabel('Product'); plt.ylabel('Sales')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Sales'], bins=6)
plt.title('Sales Distribution')
plt.xlabel('Sales'); plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Quantity'], df['Sales'])
plt.title('Quantity vs Sales')
plt.xlabel('Quantity'); plt.ylabel('Sales')
plt.show()
```

## 10. Business Insights

```python
top_product = product_analysis.index[0]
top_region = region_analysis.index[0]
print(f'Top product by sales: {top_product} ({product_analysis.iloc[0].Total_Sales:,.0f})')
print(f'Top region by sales: {top_region} ({region_analysis.iloc[0].Total_Sales:,.0f})')
print(f'Overall sales: {df.Sales.sum():,.0f}')
```
